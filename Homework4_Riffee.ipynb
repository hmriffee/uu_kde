```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
X = [5, 7, 7, 8, 10]
h = 1.5 
```

**Problem 3 A**

```python
def kde_estimator(sample, x, h):
    indicator_outputs = [int((value - x) <= h) for value in sample]
    return sum(indicator_outputs)/(2*len(sample)*h)
```

```python
print(f'''f_hat_h(x), at x = 6.0 = {kde_estimator(X, 6, h)},
       x = 6.5 = {kde_estimator(X, 6.5, h)}, 
       x = 7.0 = {kde_estimator(X, 7, h)}, 
       x = 8.0= {kde_estimator(X, 8, h)} 
       and x = 8.5 = {kde_estimator(X, 8.5, h)}.''')
```

**Problem 3 B**

```python
def gauss_estimator(sample, x, h):
    gauss_outputs = [(1/np.sqrt(2*np.pi))*np.exp((-1/2)*((value-x)/h)**2) for value in sample]
    return sum(gauss_outputs)/(len(sample)*h)
```

```python
print(f'''f_hat_h(x), at x = 6.0 = {gauss_estimator(X, 6, h)},
       x = 6.5 = {gauss_estimator(X, 6.5, h)}, 
       x = 7.0 = {gauss_estimator(X, 7, h)}, 
       x = 8.0= {gauss_estimator(X, 8, h)} 
       and x = 8.5 = {gauss_estimator(X, 8.5, h)}.''')
```

**Problem 3 C**

The uniform kernel estimator has big jumps between x = 6 and 6.5, and between x = 8 and 8.5. This can be attributed to the shape of the uniform kernel, which is uniform across all the bandwidth with abrupt edges that fall off to immediately to 0. So these outer values (x = 6 and 8.5) are just edging out past the boundary zone of the uniform kernels for the inner values, which causes the estimator to fall off abruptly.

In comparison, the gaussian kernel is bell-shaped around each sample value, which creates a gentle gradient in the estimator that decreases the further away you get from the observed value. Taken together, this leads to a gradient among the sum of the kernels, and does not produce the sharp cliff-like edge that happens at the boundary of several overlapping uniform kernel estimators. 

**Problem 3 D**

As  more sample data is put into your uniform KDE, and as its bandwidth decreases, thereby decreasing its bias, the uniform KDE begins to approximate a smooth curve! In contrast, the gaussian is smooth from the start. The smooth curve is a better reflection of real life, in which I think that most trends and relationships are not blocky and stepwise. In practice, I would probably prefer the gaussian KDE, as the smooth gradients it produces are better approximations of real life.

However, the uniform density estimator is easier to calculate and compute and interpret, so in some cases it may be better from a simplicity standpoint.
